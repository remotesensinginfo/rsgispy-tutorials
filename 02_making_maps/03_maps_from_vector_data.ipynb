# Maps from Vector Data Layers

*Section 2 — Making maps, notebook 3 of 4.* Previous: `02_maps_from_eo_data.ipynb`. Next: `04_combining_data_sources.ipynb`.

**You will learn:**
- how to read vector layers with GeoPandas and map them with `rsgispy.mapping.vector.create_vec_lyr_map`;
- how to change fill and line colours, line widths and transparency, and overlay several layers;
- how to add a background map (OpenStreetMap / CartoDB tiles) with `rsgispy.mapping.basemaps.add_contextily_basemap`;
- how to show vector maps side by side with `rsgispy.mapping.layout.create_map_panels`;
- how to build a multi-panel figure with labelled maps with `gridspec`.

**Before you start:** the matplotlib basics from `01_intro_matplotlib.ipynb` and the image maps from `02_maps_from_eo_data.ipynb`.

**Data:** `../data/gmw/gmw_v3_lyrs_sub.gpkg` (Global Mangrove Watch v3 extents, 1996–2020, for a region of Venezuela). The background maps are downloaded from the internet, so you need a network connection.

**Run time:** less than 1 minute.

It is common that we also want to produce maps using vector layers and we will now look at the functions rsgispy provides for that.

When dealing with vector data we will load the data using the geopandas (https://geopandas.org) module and pass the vector data to the rsgispy functions as geopandas dataframes.

## Import Modules

```python
%matplotlib inline
import os

import geopandas
import matplotlib.pyplot as plt
import rsgispy.mapping.basemaps
import rsgispy.mapping.elements
import rsgispy.mapping.layout as map_layout
import rsgispy.mapping.vector

# The figures saved by this notebook are written to the outputs directory
out_dir = "outputs"
os.makedirs(out_dir, exist_ok=True)
```

## Specify Input File Names

```python
vec_file = "../data/gmw/gmw_v3_lyrs_sub.gpkg"
lyr_1996 = "mng_mjr_1996"
lyr_2007 = "mng_mjr_2007"
lyr_2008 = "mng_mjr_2008"
lyr_2009 = "mng_mjr_2009"
lyr_2010 = "mng_mjr_2010"
lyr_2015 = "mng_mjr_2015"
lyr_2016 = "mng_mjr_2016"
lyr_2017 = "mng_mjr_2017"
lyr_2018 = "mng_mjr_2018"
lyr_2019 = "mng_mjr_2019"
lyr_2020 = "mng_mjr_2020"
```

```python
gmw_1996_gdf = geopandas.read_file(vec_file, layer=lyr_1996)
```

Note, you can view the contents of the geopandas dataframe within a notebook. So if you want to check what attributes the vector layer has or something like that - although in this case there isn't much information attached.

## Create a Map using a Vector Layer

The process is largely the same as with raster datasets, in that the element rendering the map is matplotlib and the rsgispy function is just adding information to the matplotlib plot:

```python
bbox = [-62.46, -62.25, 9.86, 10.05]

# Create a plot using the matplotlib library
fig, ax = plt.subplots(figsize=(5, 5))

# The function returns a list of the matplotlib collections it has drawn.
# Assigning it to a variable stops the notebook printing it; below, where
# we do not need it, it is assigned to `_`.
vec_colls = rsgispy.mapping.vector.create_vec_lyr_map(ax, gmw_1996_gdf, bbox)
```

The x-axis coordinates are overlapping one another which isn't very help, so lets sort that out:

```python
bbox = [-62.46, -62.25, 9.86, 10.05]

# Create a plot using the matplotlib library
fig, ax = plt.subplots(figsize=(5, 5))

_ = rsgispy.mapping.vector.create_vec_lyr_map(ax, gmw_1996_gdf, bbox)
ax.tick_params(axis="x", labelrotation=45)
```

## Let's change the colour of the polygons:

```python
bbox = [-62.46, -62.25, 9.86, 10.05]

# Create a plot using the matplotlib library
fig, ax = plt.subplots(figsize=(5, 5))

_ = rsgispy.mapping.vector.create_vec_lyr_map(ax, gmw_1996_gdf, bbox, vec_fill_clrs="green")
ax.tick_params(axis="x", labelrotation=45)
```

```python
bbox = [-62.46, -62.25, 9.86, 10.05]

# Create a plot using the matplotlib library
fig, ax = plt.subplots(figsize=(5, 5))

_ = rsgispy.mapping.vector.create_vec_lyr_map(
    ax,
    gmw_1996_gdf,
    bbox,
    vec_fill_clrs="#5cc460",
    vec_line_widths=0,  # defining the line with as 0 removes outlines
)
ax.tick_params(axis="x", labelrotation=45)
```

## Only Show Outlines

Now here we need to get geopandas to help us out, using the `boundary` option on a polygons vector layer returns a vector layer with just lines and therefore only the outlines will be displayed: 

```python
bbox = [-62.46, -62.25, 9.86, 10.05]

# Create a plot using the matplotlib library
fig, ax = plt.subplots(figsize=(5, 5))

_ = rsgispy.mapping.vector.create_vec_lyr_map(
    ax,
    gmw_1996_gdf.boundary,
    bbox,
    vec_line_clrs="black",
    vec_line_widths=1,
)
ax.tick_params(axis="x", labelrotation=45)
```

## Multiple Layers on the Same Map

The `rsgispy.mapping.vector.create_vec_lyr_map` function also allows a list of geopandas dataframes to be passed and it will overlay those layers. 

First we need to open another GMW layer:

```python
gmw_2020_gdf = geopandas.read_file(vec_file, layer=lyr_2020)
```

We can pass a list of geopandas dataframes but we need to also specify a list of colours so we can see the different layers:

Note, the order of the list will be the order in which the layers will be rendereds so in this same mangroves have expanded between 1996 and 2020 so we want to display the 2020 first and then the 1996 on top:

```python
bbox = [-62.46, -62.25, 9.86, 10.05]

# Create a plot using the matplotlib library
fig, ax = plt.subplots(figsize=(5, 5))

_ = rsgispy.mapping.vector.create_vec_lyr_map(
    ax,
    [gmw_2020_gdf, gmw_1996_gdf],
    bbox,
    vec_fill_clrs=["#5A95D8", "#5CC460"],
    vec_line_widths=0,
)
ax.tick_params(axis="x", labelrotation=45)
```

Below just illustrates providing the list of layers in both orders - you can then see both gain and loss between the two orders. The two maps are drawn side by side with `create_map_panels`, which was introduced in the previous notebook:

```python
bbox = [-62.46, -62.25, 9.86, 10.05]

# Draw the two maps side by side with create_map_panels (see the
# previous notebook): a "vector" panel takes a GeoDataFrame, or a list
# of them, as vec, and the options of create_vec_lyr_map in map_kw.
fig, axs = map_layout.create_map_panels(
    [
        map_layout.MapPanel(
            "vector",
            vec=[gmw_2020_gdf, gmw_1996_gdf],
            title="2020, 1996",
            map_kw={"vec_fill_clrs": ["#5A95D8", "#5CC460"], "vec_line_widths": 0},
        ),
        map_layout.MapPanel(
            "vector",
            vec=[gmw_1996_gdf, gmw_2020_gdf],
            title="1996, 2020",
            map_kw={"vec_fill_clrs": ["#5CC460", "#5A95D8"], "vec_line_widths": 0},
        ),
    ],
    bbox=bbox,
)
```

## Background Map - contextily

It is often useful to have some background mapping behind the vector layer, which should be some satellite imagery using the `rsgispy.mapping.raster.create_raster_img_map` function but we can also use other layers such as open street map and the contextily module (https://contextily.readthedocs.io) allows use to easily add such layers:


A list of all the layers available is here: https://contextily.readthedocs.io/en/latest/providers_deepdive.html

```python
# We first need to import the contextily module
import contextily
```

### Using the topographic open street map render:

```python
bbox = [-62.46, -62.25, 9.86, 10.05]

# Create a plot using the matplotlib library
fig, ax = plt.subplots(figsize=(5, 5))

_ = rsgispy.mapping.vector.create_vec_lyr_map(
    ax,
    gmw_2020_gdf,
    bbox,
    vec_fill_clrs="#5CC460",
    vec_line_widths=0,
)
ax.tick_params(axis="x", labelrotation=45)

_ = rsgispy.mapping.basemaps.add_contextily_basemap(
    ax, epsg_crs=4326, cx_src=contextily.providers.OpenTopoMap
)
```

### Changing the Copyright text

Sometimes the default contextily copyright string can be very long and make the map look untidy, we can therefore edit it:

```python
bbox = [-62.46, -62.25, 9.86, 10.05]

# Create a plot using the matplotlib library
fig, ax = plt.subplots(figsize=(5, 5))

_ = rsgispy.mapping.vector.create_vec_lyr_map(
    ax,
    gmw_2020_gdf,
    bbox,
    vec_fill_clrs="#5CC460",
    vec_line_widths=0,
)
ax.tick_params(axis="x", labelrotation=45)

_ = rsgispy.mapping.basemaps.add_contextily_basemap(
    ax,
    epsg_crs=4326,
    cx_src=contextily.providers.OpenTopoMap,
    cx_attribution="Copyright OpenStreetMap - background map",
)
```

Or remove it and put the copyright infomation to a figure caption:

```python
bbox = [-62.46, -62.25, 9.86, 10.05]

# Create a plot using the matplotlib library
fig, ax = plt.subplots(figsize=(5, 5))

_ = rsgispy.mapping.vector.create_vec_lyr_map(
    ax,
    gmw_2020_gdf,
    bbox,
    vec_fill_clrs="#5CC460",
    vec_line_widths=0,
)
ax.tick_params(axis="x", labelrotation=45)

_ = rsgispy.mapping.basemaps.add_contextily_basemap(
    ax, epsg_crs=4326, cx_src=contextily.providers.OpenTopoMap, cx_attribution=False
)
```

### Multiple Maps in Figure:

```python
# For using gridspec the module needs to be importing
import matplotlib.gridspec as gridspec
```

While this is quite a lot of code below, note that it is largely duplicated from year to year - this probably indicates that this could be produced using a loop to reduce code duplication... The code below shows how to build such a figure yourself with `gridspec`, which gives you full control over the layout; `create_map_panels` (used above) is a quicker way to put maps side by side.

Note, the `rsgispy.mapping.elements.add_axis_label` function has been used to label the individual maps so they can be referenced in a document etc. 

```python
# Create the figure and gridspec
fig = plt.figure(figsize=(15, 20))
gspec = gridspec.GridSpec(ncols=3, nrows=4, figure=fig)

#################################################
# 1996
ax96 = fig.add_subplot(gspec[0, 0])
_ = rsgispy.mapping.vector.create_vec_lyr_map(
    ax96,
    gmw_1996_gdf,
    bbox,
    vec_fill_clrs="#009900",
    vec_line_widths=0,
    vec_fill_alphas=0.5,
)
_ = rsgispy.mapping.basemaps.add_contextily_basemap(
    ax96,
    epsg_crs=4326,
    cx_src=contextily.providers.Esri.WorldGrayCanvas,
    cx_attribution=False,
)
ax96.set_title("1996")
_ = rsgispy.mapping.elements.add_axis_label(ax96, lbl_text="a)")
#################################################

#################################################
# 2007
ax07 = fig.add_subplot(gspec[0, 1])
_ = rsgispy.mapping.vector.create_vec_lyr_map(
    ax07,
    gmw_2007_gdf,
    bbox,
    vec_fill_clrs="#acc0dd",
    vec_line_widths=0,
    vec_fill_alphas=0.5,
)
_ = rsgispy.mapping.basemaps.add_contextily_basemap(
    ax07,
    epsg_crs=4326,
    cx_src=contextily.providers.Esri.WorldGrayCanvas,
    cx_attribution=False,
)
ax07.set_title("2007")
_ = rsgispy.mapping.elements.add_axis_label(ax07, lbl_text="b)")
#################################################

#################################################
# 2008
ax08 = fig.add_subplot(gspec[0, 2])
_ = rsgispy.mapping.vector.create_vec_lyr_map(
    ax08,
    gmw_2008_gdf,
    bbox,
    vec_fill_clrs="#91b5d6",
    vec_line_widths=0,
    vec_fill_alphas=0.5,
)
_ = rsgispy.mapping.basemaps.add_contextily_basemap(
    ax08,
    epsg_crs=4326,
    cx_src=contextily.providers.Esri.WorldGrayCanvas,
    cx_attribution=False,
)
ax08.set_title("2008")
_ = rsgispy.mapping.elements.add_axis_label(ax08, lbl_text="c)")
#################################################

#################################################
# 2009
ax09 = fig.add_subplot(gspec[1, 0])
_ = rsgispy.mapping.vector.create_vec_lyr_map(
    ax09,
    gmw_2009_gdf,
    bbox,
    vec_fill_clrs="#73a9cf",
    vec_line_widths=0,
    vec_fill_alphas=0.5,
)
_ = rsgispy.mapping.basemaps.add_contextily_basemap(
    ax09,
    epsg_crs=4326,
    cx_src=contextily.providers.Esri.WorldGrayCanvas,
    cx_attribution=False,
)
ax09.set_title("2009")
_ = rsgispy.mapping.elements.add_axis_label(ax09, lbl_text="d)")
#################################################

#################################################
# 2010
ax10 = fig.add_subplot(gspec[1, 1])
_ = rsgispy.mapping.vector.create_vec_lyr_map(
    ax10,
    gmw_2010_gdf,
    bbox,
    vec_fill_clrs="#509ac6",
    vec_line_widths=0,
    vec_fill_alphas=0.5,
)
_ = rsgispy.mapping.basemaps.add_contextily_basemap(
    ax10,
    epsg_crs=4326,
    cx_src=contextily.providers.Esri.WorldGrayCanvas,
    cx_attribution=False,
)
ax10.set_title("2010")
_ = rsgispy.mapping.elements.add_axis_label(ax10, lbl_text="e)")
#################################################

#################################################
# 2015
ax15 = fig.add_subplot(gspec[1, 2])
_ = rsgispy.mapping.vector.create_vec_lyr_map(
    ax15,
    gmw_2015_gdf,
    bbox,
    vec_fill_clrs="#2f8bbe",
    vec_line_widths=0,
    vec_fill_alphas=0.5,
)
_ = rsgispy.mapping.basemaps.add_contextily_basemap(
    ax15,
    epsg_crs=4326,
    cx_src=contextily.providers.Esri.WorldGrayCanvas,
    cx_attribution=False,
)
ax15.set_title("2015")
_ = rsgispy.mapping.elements.add_axis_label(ax15, lbl_text="f)")
#################################################


#################################################
# 2016
ax16 = fig.add_subplot(gspec[2, 0])
_ = rsgispy.mapping.vector.create_vec_lyr_map(
    ax16,
    gmw_2016_gdf,
    bbox,
    vec_fill_clrs="#1379b5",
    vec_line_widths=0,
    vec_fill_alphas=0.5,
)
_ = rsgispy.mapping.basemaps.add_contextily_basemap(
    ax16,
    epsg_crs=4326,
    cx_src=contextily.providers.Esri.WorldGrayCanvas,
    cx_attribution=False,
)
ax16.set_title("2016")
_ = rsgispy.mapping.elements.add_axis_label(ax16, lbl_text="g)")
#################################################

#################################################
# 2017
ax17 = fig.add_subplot(gspec[2, 1])
_ = rsgispy.mapping.vector.create_vec_lyr_map(
    ax17,
    gmw_2017_gdf,
    bbox,
    vec_fill_clrs="#0569a5",
    vec_line_widths=0,
    vec_fill_alphas=0.5,
)
_ = rsgispy.mapping.basemaps.add_contextily_basemap(
    ax17,
    epsg_crs=4326,
    cx_src=contextily.providers.Esri.WorldGrayCanvas,
    cx_attribution=False,
)
ax17.set_title("2017")
_ = rsgispy.mapping.elements.add_axis_label(ax17, lbl_text="h)")
#################################################

#################################################
# 2018
ax18 = fig.add_subplot(gspec[2, 2])
_ = rsgispy.mapping.vector.create_vec_lyr_map(
    ax18,
    gmw_2018_gdf,
    bbox,
    vec_fill_clrs="#045d92",
    vec_line_widths=0,
    vec_fill_alphas=0.5,
)
_ = rsgispy.mapping.basemaps.add_contextily_basemap(
    ax18,
    epsg_crs=4326,
    cx_src=contextily.providers.Esri.WorldGrayCanvas,
    cx_attribution=False,
)
ax18.set_title("2018")
_ = rsgispy.mapping.elements.add_axis_label(ax18, lbl_text="i)")
#################################################

#################################################
# 2019
ax19 = fig.add_subplot(gspec[3, 0])
_ = rsgispy.mapping.vector.create_vec_lyr_map(
    ax19,
    gmw_2019_gdf,
    bbox,
    vec_fill_clrs="#034b76",
    vec_line_widths=0,
    vec_fill_alphas=0.5,
)
_ = rsgispy.mapping.basemaps.add_contextily_basemap(
    ax19,
    epsg_crs=4326,
    cx_src=contextily.providers.Esri.WorldGrayCanvas,
    cx_attribution=False,
)
ax19.set_title("2019")
_ = rsgispy.mapping.elements.add_axis_label(ax19, lbl_text="j)")
#################################################

#################################################
# 2020
ax20 = fig.add_subplot(gspec[3, 1])
_ = rsgispy.mapping.vector.create_vec_lyr_map(
    ax20,
    gmw_2020_gdf,
    bbox,
    vec_fill_clrs="#023858",
    vec_line_widths=0,
    vec_fill_alphas=0.5,
)
_ = rsgispy.mapping.basemaps.add_contextily_basemap(
    ax20,
    epsg_crs=4326,
    cx_src=contextily.providers.Esri.WorldGrayCanvas,
    cx_attribution=False,
)
ax20.set_title("2020")
_ = rsgispy.mapping.elements.add_axis_label(ax20, lbl_text="k)")
#################################################

plt.savefig(os.path.join(out_dir, "gmw_timeseries.png"), dpi=400)
```
